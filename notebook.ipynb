```python
import os
import pandas as pd
import numpy as np
from pybaseball import batting_stats_bref
```

```python
batting_2002_2022_200pa = pandas.read_csv("batting.csv", index_col=0)
batting_2002_2022_200pa.head()

```

```python
batting_2002_2022_200pa = batting_2002_2022_200pa.groupby("IDfg", group_keys = False).filter(lambda x: x.shape[0] > 1)
```

```python
batting_2002_2022_200pa
```

```python
def next_season(player):
    player = player.sort_values("Season")
    player["Next_WAR"] = player["WAR"].shift(-1)
    return player

batting_2002_2022_200pa = batting_2002_2022_200pa.sort_values(["IDfg", "Season"])
batting_2002_2022_200pa["Next_WAR"] = batting_2002_2022_200pa.groupby("IDfg")["WAR"].shift(-1)
```

```python
batting_2002_2022_200pa[["Name", "Season", "WAR", "Next_WAR"]]
null_count = batting_2002_2022_200pa.isnull().sum()
null_count
cols_with_no_null = list(batting_2002_2022_200pa.columns[null_count == 0])
```

```python
batting_2002_2022_200pa = batting_2002_2022_200pa[cols_with_no_null + ["Next_WAR"]].copy()
batting_2002_2022_200pa
```

```python
batting_2002_2022_200pa.dtypes[batting_2002_2022_200pa.dtypes == "object"]
del batting_2002_2022_200pa["Age Rng"]
del batting_2002_2022_200pa["Dol"]
```

```python
batting_2002_2022_200pa["Team_Code"] = batting_2002_2022_200pa["Team"].astype("category").cat.codes
```

```python
batting_full = batting_2002_2022_200pa.copy()
batting_2002_2022_200pa = batting_2002_2022_200pa.dropna().copy()
```

```python
from sklearn.linear_model import Ridge
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.model_selection import TimeSeriesSplit

rr = Ridge(alpha = 1)

split = TimeSeriesSplit(n_splits=3)

sfs = SequentialFeatureSelector(rr, n_features_to_select=20, direction="forward", cv=split, n_jobs=4)
```

```python
removed_columns = ["Next_WAR", "Name", "Team", "IDfg", "Season"]
selected_columns = batting_2002_2022_200pa.columns[~batting_2002_2022_200pa.columns.isin(removed_columns)]
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
batting_2002_2022_200pa[selected_columns] = scaler.fit_transform(batting_2002_2022_200pa[selected_columns])
```

```python
batting_2002_2022_200pa.describe()
```

```python
sfs.fit(batting_2002_2022_200pa[selected_columns], batting_2002_2022_200pa["Next_WAR"])
```

```python
predictors = list(selected_columns[sfs.get_support()])
```

```python
def backtest(data, model, predictors, start = 5, step = 1):
    all_predictions = []
    
    years = sorted(data["Season"].unique())
    
    for i in range(start, len(years), step):
        curr_year = years[i]
        
        training_set = data[data["Season"] < curr_year]
        testing_set = data[data["Season"] == curr_year]
        
        model.fit(training_set[predictors], training_set["Next_WAR"])
        
        preds = model.predict(testing_set[predictors])
        preds = pd.Series(preds, index = testing_set.index)
        
        combined = pd.concat([testing_set["Next_WAR"], preds], axis = 1)
        combined.columns = ["Actual_WAR", "Pred_WAR"]
        all_predictions.append(combined)
    return pd.concat(all_predictions)
```

```python
predictions = backtest(batting_2002_2022_200pa, rr, predictors)
```

```python
predictions
```

```python
from sklearn.metrics import mean_squared_error

mean_squared_error(predictions["Actual_WAR"], predictions["Pred_WAR"])
```

```python
batting_2002_2022_200pa = batting_2002_2022_200pa.sort_values(["IDfg", "Season"]).copy()
batting_2002_2022_200pa["player_season"] = batting_2002_2022_200pa.groupby("IDfg").cumcount()

def war_corr(group):
    return group[["player_season", "WAR"]].expanding().corr().loc[(slice(None), "player_season"), "WAR"]

batting_2002_2022_200pa["war_corr"] = batting_2002_2022_200pa.groupby("IDfg", group_keys=False).apply(war_corr, include_groups=False).values
batting_2002_2022_200pa["war_corr"] = batting_2002_2022_200pa["war_corr"].fillna(0)

batting_2002_2022_200pa["war_diff"] = batting_2002_2022_200pa["WAR"] / batting_2002_2022_200pa.groupby("IDfg")["WAR"].shift(1)
batting_2002_2022_200pa["war_diff"] = batting_2002_2022_200pa["war_diff"].fillna(1)
batting_2002_2022_200pa.loc[np.isinf(batting_2002_2022_200pa["war_diff"]), "war_diff"] = 1 
```

```python
def group_averages(df):
    return df["WAR"] / df["WAR"].mean()

batting_2002_2022_200pa["war_season"] = batting_2002_2022_200pa.groupby("Season", group_keys=False).apply(group_averages)

new_predictors = predictors + ["player_season", "war_corr", "war_season", "war_diff"]

predictions = backtest(batting_2002_2022_200pa, rr, new_predictors)

mean_squared_error(predictions["Actual_WAR"], predictions["Pred_WAR"])
```
